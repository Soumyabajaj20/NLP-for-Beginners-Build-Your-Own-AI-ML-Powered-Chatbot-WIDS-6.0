## Section 3: Strings, the Heart of NLP ❤️
📖 *[W3Schools Strings](https://www.w3schools.com/python/python_strings.asp) · [String Methods](https://www.w3schools.com/python/python_strings_methods.asp)*

Every chatbot, translator and spam filter starts by **manipulating strings**. Master these methods and you're 30% of the way to NLP preprocessing!

```python
text = "  I LOVED this movie! The acting was AMAZING.  "

print(text.lower())          # all lowercase
print(text.upper())          # all uppercase
print(text.strip())          # remove spaces at the start/end
print(text.replace("!", "")) # replace characters
print(len(text))             # number of characters
```

### Splitting text into words: your first **tokenizer**!
`split()` breaks a string into a **list of words**. In NLP, this is called **tokenization**.

```python
sentence = "natural language processing is fun"
words = sentence.split()
print(words)
print("Number of words:", len(words))

# join() does the opposite
print("-".join(words))
```

### Indexing & slicing
Each character has a position (index) starting from **0**. Negative indexes count from the end.

```python
word = "transformer"
print(word[0])      # first character
print(word[-1])     # last character
print(word[0:5])    # characters 0 to 4 -> "trans"
print(word[::-1])   # reversed!

print("form" in word)   # check if a substring exists -> True
```

## Section 4: Loops 🔁
📖 *[W3Schools For Loops](https://www.w3schools.com/python/python_for_loops.asp)*

A `for` loop repeats code for each item in a sequence. A `while` loop repeats as long as a condition is True.

```python
words = ["I", "love", "NLP"]
for w in words:
    print(w, "->", len(w), "letters")

# range(n) gives 0, 1, ..., n-1
for i in range(3):
    print("Iteration", i)

# loop over characters of a string
vowels = 0
for ch in "machine learning":
    if ch in "aeiou":
        vowels += 1          # same as vowels = vowels + 1
print("Vowels:", vowels)
```

```python
# while loop
count = 3
while count > 0:
    print(count)
    count -= 1
print("Liftoff! 🚀")
```

## Section 5: Lists 📋
🎥 *[Corey Schafer: Lists, Tuples, and Sets](https://www.youtube.com/watch?v=W8KRzm-HUcc)*

A **list** stores many items in order. In NLP, a sentence usually becomes a list of words (tokens).

```python
tokens = ["chatgpt", "is", "a", "large", "language", "model"]

print(tokens[0], tokens[-1])     # first, last
print(tokens[1:4])               # slicing works like strings
tokens.append("!")               # add at the end
print(tokens)
tokens.remove("!")               # remove an item
print(len(tokens), "tokens")
print(sorted(tokens))            # sorted copy
```

### List comprehensions: a Python superpower ⚡
A short way to build a new list from an old one. You'll see this *everywhere* in NLP code.

```python
words = ["Hello", "WORLD", "Nlp", "Rocks"]

lower_words = [w.lower() for w in words]
print(lower_words)

long_words = [w for w in words if len(w) > 4]
print(long_words)
```

### Sets: unique items only
`set()` removes duplicates. Handy for finding the **vocabulary** (unique words) of a text!

```python
text = "the cat sat on the mat the end"
vocab = set(text.split())
print(vocab)
print("Vocabulary size:", len(vocab))
```

## Section 6: Dictionaries 📖
📖 *[W3Schools Dictionaries](https://www.w3schools.com/python/python_dictionaries.asp)*

A **dictionary** maps **keys** to **values**, like a real dictionary maps words to meanings.
In NLP, the classic use is **counting word frequencies**. This idea is the basis of *bag-of-words*, which you'll learn next week!

```python
student = {"name": "Aarav", "branch": "CSE", "year": 1}
print(student["name"])
student["year"] = 2              # update
student["hostel"] = 4            # add a new key
print(student)

for key, value in student.items():
    print(key, "->", value)
```

### Counting words (THE most important pattern this week!)

```python
text = "to be or not to be that is the question"
freq = {}

for word in text.split():
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)

# Shortcut using .get(key, default)
freq2 = {}
for word in text.split():
    freq2[word] = freq2.get(word, 0) + 1
print(freq2 == freq)

# Python also has a built-in counter!
from collections import Counter
print(Counter(text.split()).most_common(3))
```

## Section 7: Functions 🧩
📖 *[W3Schools Functions](https://www.w3schools.com/python/python_functions.asp)*

A **function** is a reusable block of code. You define it once with `def` and use it many times.
`return` sends a value back to whoever called the function.

```python
def greet(name):
    return f"Hello, {name}! Welcome to NLP."

print(greet("Riya"))
print(greet("Kabir"))

def count_words(text):
    """Return the number of words in text."""
    return len(text.split())

print(count_words("functions make code reusable"))

# default arguments
def shout(text, times=1):
    return (text.upper() + "! ") * times

print(shout("nlp"))
print(shout("nlp", times=3))
```

## Section 8: Mini Project 🛠️ Text Analyser
Let's combine everything into a small tool that analyses any piece of text. Read the code carefully. Every line uses something you learned today!

```python
import string

def analyse_text(text):
    # 1. clean: lowercase + remove punctuation
    clean = text.lower()
    for p in string.punctuation:          # string.punctuation = '!"#$%&...'
        clean = clean.replace(p, "")

    # 2. tokenize
    tokens = clean.split()

    # 3. count frequencies
    freq = {}
    for t in tokens:
        freq[t] = freq.get(t, 0) + 1

    # 4. find the most common word
    top_word = max(freq, key=freq.get)

    return {
        "characters": len(text),
        "words": len(tokens),
        "unique_words": len(freq),
        "most_common_word": top_word,
        "avg_word_length": round(sum(len(t) for t in tokens) / len(tokens), 2),
    }

sample = "AI is changing the world. AI writes code, AI translates languages, and AI chats with us!"
for k, v in analyse_text(sample).items():
    print(f"{k:>18}: {v}")
```

## Section 9: Intro to pandas 🐼
🎥 *[NumPy tutorial](https://www.youtube.com/watch?v=QUT1VHiLmmI) · [pandas tutorial](https://www.youtube.com/watch?v=vmEHCJofslg)*

Real NLP datasets (reviews, tweets, emails) come as **tables**. `pandas` is the Python library for working with tables, which it calls **DataFrames**.

```python
import pandas as pd

data = {
    "review": ["Absolutely loved it!", "Worst purchase ever", "It was okay, nothing special",
               "Fantastic quality and fast delivery", "Terrible, broke in two days"],
    "rating": [5, 1, 3, 5, 1],
}
df = pd.DataFrame(data)
df
```

```python
print(df.shape)            # (rows, columns)
print(df["rating"].mean()) # average rating
df.head(3)                 # first 3 rows
```

### Adding new columns with `.apply()`
`.apply(function)` runs a function on **every row** of a column. This is how we'll preprocess thousands of texts at once.

```python
df["num_words"] = df["review"].apply(lambda t: len(t.split()))
df["label"] = df["rating"].apply(lambda r: "positive" if r >= 4 else ("negative" if r <= 2 else "neutral"))
df
```

```python
# Filtering rows
print(df[df["label"] == "positive"])

# Counting labels
print(df["label"].value_counts())

# Reading a real CSV from the internet works the same way:
# df = pd.read_csv("https://some-url/data.csv")
```
